```python
llm = ChatOpenAI(model_name="gpt-3.5-turbo-1106")

learn_template= """
I want you to acts as consultant for AI trainings
Return a list of topics and why ot is important to learn in given area of AI
The description should be relevant to recent advancement in AI
What are some good topics to learn in {AI_topic}
"""

prompt_template = PromptTemplate(
    input_variables=["AI_topic"],
    template =learn_template
)

description = "Please suggest me good topics to learn for Agentic AI"

formatted_template = prompt_template.format(AI_topic=description)
chain = LLMChain(llm=llm, prompt=prompt_template)
print(chain.run(description))
```

```python
# Define simple tool

def weather_tool_function(query: str) -> str:
    return f"Tool Response : {query}"

# Creating the tool from function
weather_tool = Tool.from_function(func=weather_tool_function, name="simple_tool", description="A Simple tool")
#ddg_search = DuckDuckGoSearchRun()

# Initialize  agent
tools = [ weather_tool]

# Create agent
agent = initialize_agent(tools, llm=llm, agent_type=AgentType.ZERO_SHOT_REACT_DESCRIPTION, verbose= True)

# Use agent
response = agent.run("Whats the weather like today in Pune?")
print(response)
    
```

## Agent 2

```python
prompt_template = "Summarize the following content : {content}"
#ddg_search = DuckDuckGoSearchRun()

llm = ChatOpenAI(model="gpt-3.5-turbo-16k")
llm_chain = LLMChain(llm=llm,
                    prompt= PromptTemplate.from_template(prompt_template)
                    )

summarize_tool = Tool.from_function(
    func=llm_chain.run,
    name="Summarize",
    description="Summarize web page"
)

tools = [summarize_tool]

agent2 = initialize_agent(tools=tools,
                          agent_type=AgentType.ZERO_SHOT_REACT_DESCRIPTION, 
                         llm = llm,
                         verbose= True,
                         handle_parsing_erros= True)

prompt = """ Please tell me how to become AI Expert"""

print(agent2.invoke(prompt))
```
